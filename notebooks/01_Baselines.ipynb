```python
import pandas as pd
import sys
```

```python
sys.path.append(r"..\src\models")   

from baselines.popularity import (
    compute_movie_stats,
    top_avg_rating,
    top_weighted_rating,
    top_weighted_rating_tmdb
)
```

```python
movies = pd.read_csv(r"..\data\processed\movies_clean.csv")
ratings = pd.read_csv(r"..\data\processed\ratings_clean.csv")
```

```python
movie_stats = compute_movie_stats(ratings, movies)
movie_stats.head()
```

```python
# baseline 1. plain avg. 
top_avg = top_avg_rating(movie_stats)
print("Top 10 by Average Rating:")
print(top_avg)
```

Plain averages overrate movies with very few votes.

```python
# baseline 2. IMDb WR
top_wr = top_weighted_rating(movie_stats)
print("Top 10 by IMDb Weighted Rating(based on ratings dataset):")
print(top_wr)
```

```python
# baseline 2. IMDb WR (using TMDb votes)
top_wr_tmdb = top_weighted_rating_tmdb(movies)
print("Top 10 by IMDb Weighted Rating(based on votes from TMDb):")
print(top_wr_tmdb)
```

WR corrects this by blending with the global mean.
