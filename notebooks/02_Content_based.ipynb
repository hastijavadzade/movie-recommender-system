```python
import pandas as pd
import sys
from sklearn.decomposition import TruncatedSVD
import numpy as np
```

```python
movies = pd.read_csv(r"..\data\processed\movies_clean.csv")
# ratings = pd.read_csv(r"..\data\processed\ratings_clean.csv")
ratings_small = pd.read_csv(r"..\data\raw\ratings_small.csv")
```

```python
sys.path.append(r"..\src\models")   

from cb.content_based import (
    compute_tf_matrix,
    compute_cv_matrix,
    compute_truncatedSVD_matrix,
    recommend_movie
)
```

```python
tf_matrix = compute_tf_matrix(movies)
```

```python
# saving the tf matrix
from scipy.sparse import save_npz

save_npz(r"..\src\models\cb\tf_matrix.npz", tf_matrix)
```

```python
tf_matrix.shape
```

```python
movies['title_lower'] = movies['original_title'].str.lower() # for case-insensitive matching
```

```python
recommended_movies = recommend_movie(movies, "tangled", tf_matrix)

print("Recommended movies for Tangled\n")
top_movies = movies[movies['movieId'].isin(recommended_movies)][
        ["movieId", "original_title", "genres", "cast", "director"]
    ]

    # Keep the order the recommender gave
top_movies = top_movies.set_index("movieId").loc[recommended_movies].reset_index()
top_movies
```

```python
cv_matrix = compute_cv_matrix(movies)
```

```python
recommended_movies = recommend_movie(movies, "Tangled", cv_matrix)

print("Recommended movies for Tangled\n")
top_movies = movies[movies['movieId'].isin(recommended_movies)][
        ["movieId", "original_title", "genres", "cast", "director"]
    ]

    # Keep the order the recommender gave
top_movies = top_movies.set_index("movieId").loc[recommended_movies].reset_index()
top_movies
```
