```python
import pandas as pd
import sys
```

```python
movies = pd.read_csv(r"..\data\processed\movies_clean.csv")
ratings_small = pd.read_csv(r"..\data\raw\ratings_small.csv")
```

item-item k-NN

```python
sys.path.append(r"..\src\models")   

from cf.knn_cf import (
    build_user_item_matrix,
    compute_item_similarity,
    recommend_item_knn
)
```

```python
# build matrix
user_item_matrix = build_user_item_matrix(ratings_small)

# compute similarity (top-100 neighbours per movie)
topk_sim = compute_item_similarity(user_item_matrix, k=100)
```

```python
# recommend for user 1
recs = recommend_item_knn(1, ratings_small, topk_sim, 10, movies)
print("Recommendations for User 1:")
for title, score in recs:
    print(f"{title} (score={score:.3f})")
```

```python
# recommend for user 10
recs = recommend_item_knn(10, ratings_small, topk_sim, 10, movies)
print("Recommendations for User 10:") # user 10 has rated more movies than user 1, thus their scores are higher
for title, score in recs:
    print(f"{title} (score={score:.3f})")
```

model based MF

```python
sys.path.append(r"..\src\models")   

from cf.mf_cf import (
    train_mf,
    recommend_mf
)
```

```python
svd = train_mf(ratings_small)
```

```python
import pickle
# save the mf model
with open(r"..\src\models\cf\mf_model.pkl", "wb") as f:
    pickle.dump(svd, f)
```

```python
recs = recommend_mf(svd, user_id=1, movies_df=movies, ratings_df=ratings_small, n=10)

print("Recommendations for User 1:")
print(recs[['original_title', 'pred_rating']])
```

```python
recs = recommend_mf(svd, user_id=10, movies_df=movies, ratings_df=ratings_small, n=10)

print("Recommendations for User 10:")
print(recs[['original_title', 'pred_rating']])
```
